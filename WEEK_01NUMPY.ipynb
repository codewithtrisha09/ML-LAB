Q3 Find the positions of:
elements in x where its value is more than its corresponding element in y, and elements in x where its value is
equals to its corresponding element in y.

```python
import numpy as np

x = np.array([21, 64, 86, 22, 74, 55, 81, 79, 90, 89])
y = np.array([21, 7, 3, 45, 10, 29, 55, 4, 37, 18])
greater = np.where(x > y)
equal = np.where(x == y)

print(greater)
print(equal)
```

Q2. Consider the following matrices :
A= ((1, 2, 3), (4, 5, 6), (7, 8, 10)) and B = ((7, 8, 10) ,(4, 5, 6), (1, 2, 3))
Write a python program to perform the following using Numeric Python (numpy).
i) Add and Subtract of the Matrix A and B, print the resultant matrix C for add and E for subtract.
ii) Compute the sum of all elements of Matrix A, sum of each column of Matrix B and sum of each row of Matrix C
iii) Product of two matrices A and B, and print the resultant matrix D
iv) Sort the elements of resultant matrix C and print the resultant Matrix E.
v) Transpose the Matrix E and print the result

```python
import numpy as np

A = np.array([[1, 2, 3],
              [4, 5, 6],
              [7, 8, 10]])

B = np.array([[7, 8, 10],
              [4, 5, 6],
              [1, 2, 3]])

C = A + B
E_sub = A - B
print("Matrix C (Addition):\n", C)
print("Matrix E (Subtraction):\n", E_sub)
```

```python
print("\n ii) Matrix Sums ")
sum_A = np.sum(A)
sum_col_B = np.sum(B, axis=0)
sum_row_C = np.sum(C, axis=1)
print("Sum of all elements of Matrix A:", sum_A)
print("Sum of each column of Matrix B:", sum_col_B)
print("Sum of each row of Matrix C:", sum_row_C)
```

```python
print("\n(iii) Matrix Multiplication")
D = np.dot(A, B)
print("Matrix D (Product of A and B):\n", D)
```

```python
print("\n (iv)Sorting")
E=np.sort(C,axis=None).reshape(C.shape)
print(" Matrix E (Sorted elements of C):\n",E)
```

```python
print("\n (v) Transpose of Matrix E")
E_transposed=E.T
print(" Transpose of E:\n",E_transposed)
```
